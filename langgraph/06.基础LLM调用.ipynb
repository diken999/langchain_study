```python
from langgraph.constants import START, END
from langgraph.graph import StateGraph
from typing import TypedDict
from langchain.chat_models import init_chat_model
from dotenv import load_dotenv

load_dotenv()

model = init_chat_model(
    model="deepseek-flash",
    extra_body={"thinking": {"type": "disabled"}}
)

class CustomState(TypedDict):
    user_message: str
    ai_message: str

def node_llm(state: CustomState):
    user_message = state.get("user_message", "hi")
    response = model.invoke(user_message)
    print("AI: ", response)
    return {"ai_message": response.text}

builder = StateGraph(CustomState)

builder.add_node("node_llm", node_llm)
builder.add_edge(START, "node_llm")
builder.add_edge("node_llm", END)

graph = builder.compile()

result = graph.invoke({"user_message": "你好👋"})
print(result)
```
