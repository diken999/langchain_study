```python
from typing import TypedDict, Annotated
from langgraph.graph import StateGraph, START, END

# 定义reducer函数
def add_reducer(old, new):
    return old + new

class CustomState(TypedDict):
    count: int
    nodes: Annotated[list[str], add_reducer]

def Node1(state: CustomState):
    print(f"Node1接收到了state: {state}")
    return {
        "count": state["count"] + 1,
        "nodes": ["Node1"]
    }

def Node2(state: CustomState):
    print(f"Node2接收到了state: {state}")
    return {
        "count": state["count"] + 1, 
        "nodes": ["Node2"]
    }

graph_builder = StateGraph(CustomState)

graph_builder.add_node("Node1", Node1)
graph_builder.add_node("Node2", Node2)
graph_builder.add_edge(START, "Node1")
graph_builder.add_edge("Node1", "Node2")
graph_builder.add_edge("Node2", END)

graph = graph_builder.compile()

# 执行graph
result = graph.invoke({"count": 0, "nodes": []})
print(result)
```
