```python
from langchain.chat_models import init_chat_model
from langchain.agents import create_agent
from langchain_core.tools import tool
from langchain.messages import HumanMessage
from dotenv import load_dotenv
from pydantic import BaseModel, Field
from typing import Literal

load_dotenv()

@tool
def square_tool(x):
    """获取一个数的平方根"""
    return x ** 0.5

# 针对比较复杂的数据结构时
class WeatherInput(BaseModel):
    location: str = Field(description="城市名或坐标")
    units: Literal["摄氏度","华氏度"] = Field(default="摄氏度", description="温度单位")
    include_forecast: bool = Field(default=False, description="是否包含未来 5 天")

@tool(args_schema=WeatherInput)
def get_weather(location: str, units: str, include_forecast: bool):
    """获取当前城市或者坐标的天气情况"""
    result = f"当前{location}的温度是22{units}"
    if include_forecast:
        result += " 未来 5 天天气晴朗"
    return result

model = init_chat_model(
    model="deepseek-flash"
)

agent = create_agent(
    model=model,
    tools=[square_tool, get_weather],
)

res_1 = agent.invoke({
    "messages": [
        HumanMessage("467 和 529 的平方根分别是多少？"),
        HumanMessage("成都天气今天怎么样？")
    ]
})

for msg in res_1["messages"]:
    print(msg.pretty_print())


```
