```python
from langchain.agents import create_agent
from langchain.agents.middleware import SummarizationMiddleware
from langgraph.checkpoint.memory import InMemorySaver
from langchain.messages import HumanMessage
from dotenv import load_dotenv

load_dotenv()

checkpointer = InMemorySaver()

sum_middleware = SummarizationMiddleware(
    model="deepseek-flash",
    # trigger=[("token", 4000)],
    trigger=("messages", 3),
    keep=("messages", 1)
)

agent = create_agent(
    model="deepseek-flash",
    middleware=[sum_middleware],
    checkpointer=checkpointer
)

config = {
    "configurable": {
        "thread_id": "thread_1"
    }
}

agent.invoke(
    { "messages": [HumanMessage("你好，我是张夏天，我是一只长毛银渐层")]},
    config
)
agent.invoke(
    { "messages": [HumanMessage("我最喜欢吃的食物是鱼罐头和冻干")] },
    config
)
agent.invoke(
    { "messages": [HumanMessage("我喜欢睡懒觉，一天的大多数时候都在睡觉")] },
    config
)

final_response = agent.invoke(
    { "messages": [HumanMessage("你还记得我吗？")] },
    config
)
print(final_response)
```
